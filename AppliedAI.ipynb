PART A – Application area review. 


Stock market prediction has been studied for decades, utilising data-driven techniques. Predicting short-horizon stock market trends can be subject to ambiguity, mainly due to the market's volatile nature. Early machine learning models like ANN were applied to stock market prediction problems, notably the work on the Tokyo stock exchange, in which neural network systems were used. 

Researchers have explored various types of AI models, as financial time series contain both noise and signal, making the signal difficult to interpret. Although many patterns exist, it is challenging to detect them because market behaviour is unstable and changes rapidly over time. This is what makes stock market prediction an active area for further research. Several factors determine whether the market will follow an upward or downward trend on a given day; for instance, any major geopolitical event can significantly affect the market. Otherwise, determining the trend of a particular stock or the market in general can be extremely noisy.  

Another factor is that prices can quickly respond to new information, making it difficult to predict short-term movements in the stock. Many papers demonstrate that short-horizon prediction is often only slightly better than chance, particularly when the goal is to indicate whether the next trading day will be up or down. This makes it crucial to evaluate the models carefully. Predicting the direction can be essential for presenting investment opportunities, which results in portfolio optimisation and informed decision-making strategies, whether to hold, buy, or sell the stock.  Even a slight improvement is considered valid, for instance, avoiding trading during uncertain periods. The aim of market analysis is not solely to make profits, but also to manage risk and reduce downside exposure/loss. 

In theory, there are two ways to design a prediction model: one using solely technical indicators and the other combining it with text-based sentiment/news.  Many factors, such as news/public opinion, can influence trend direction; however, their connection to the stock is complex, making it a challenging task. Research has tested simple trading rules and found evidence that they can capture patterns in long historical samples. 
This idea supports the approach of creating a baseline model that starts from historical price data and volume information. At the same time, adding sentiment-based data could be beneficial in conjunction with the previous price indicators. 

Technical indicators are features derived from historical stock price and volume data to predict market trends. There has been significant research in this field, particularly on whether moving average-based strategies capture random structures and evaluate technical patterns. In this project, I have tried to capture multiple horizon returns, rolling volatility, MA gap/crossover, and volume-based features for my fundamental technical price-based indicators. These indicators are often grouped into trend, momentum, volatility and volume. For instance, moving averages can represent trend direction, whereas multiple-horizon returns can represent momentum. Rolling volatility can indicate market risk conditions, and volume can help determine unusual activity. These symbols are easy to compute and interpret, making them an excellent choice for a baseline model. These indicators can be applied within a wide range of AI methods, from Logistic regression to more complex models, such as decision trees. 

While technical indicators reflect past market data, the market is subject to change due to public opinion and other company-specific news. Many studies find it resourceful to add sentiment features to capture market mood. Various online discussion platforms have also been studied to have a sentiment score. For instance, Antweiler and Frank (2004) found that Internet message channels have a statistically detectable relationship with the market. Even studying tweets by people has been explored to predict sentiment for market movement. Studying sentiment is imperative because text can capture information that historic prices alone cannot. Sentiment can have multiple sources, like social media, news, and company headlines/announcements. However, sometimes this data can be noisy and biased, leading to a weak model. Text in sentiment can be detected by using machine learning algorithms, like lexicon, to give it a positive or a negative score. In this project, sentiment, present in the form of tweets, is aggregated at a dilate level to match the stock data.

These studies add value by informing us about the impact of public sentiment and news on the market. In recent studies, the relationship between market volatility and media sentiment has been examined, as well as the role of investor attention in this relationship. Under negative sentiment conditions, increased investor attention amplifies market volatility. Overall, the literature suggests that both technical and sentiment-based indicators can be helpful to predict short-term horizon stocks. PART B – Compare and evaluate AI techniques. 

For this project, my goal is to determine whether Tesla's stock will increase or decrease on the next trading day. I have picked logistic regression for this project. I have used technical indicators and a sentiment-based indicator (tweets) to fulfil the aim of my project. 

Logistic regression is a part of a supervised machine learning technique used for binary classification. Logistic regression is a common baseline in financial prediction due to its simplicity and speed of training, producing outputs that are easy to interpret and evaluate. Logistic regression utilises the sigmoid function, which calculates the weighted sum of the input features and converts it into a probability.  Logistic regression requires numerical data, where each row consists of one trading day. In my project, I have used multiple baseline indicators, including multi-horizon returns, rolling volatility, moving average gap (to detect trends), volume change, and volume z-score. In the combined approach, I have added lagged daily mean tweet sentiment to see if the model performs any better with sentiment. 

Logistic regression is sensitive; therefore, I have done feature scaling before training the model. The expected outputs are a predicted label UP/DOWN. I have evaluated these outputs using accuracy, confusion matrices, and ROC/AUC. I have applied logistic regression twice in my project, once with only technical indicators and the second time with the addition of tweets. The main advantage of Logistic Regression is that it trains quickly and is easy to produce. It also provides a strong reference point for judging whether more complex models are worthwhile.  

Compared to Random forest, logistic regression is faster and more interpretable; however, it may not be able to capture nonlinear interactions, such as when market features depend on combinations of various indicators, like momentum behaving differently when volatility is high. Unlike neural networks, logistic regression does not require large datasets and is less likely to overfit; however, it cannot learn complex sequential dependencies.  


Neural networks for stock prediction have been one of the first techniques used for stock prediction. There are mainly two types of neural network options, Multilayer Perceptron and Long Short-Term Memory. MLP uses one row per day of engineered features, as in logistic regression/random forest. It has the ability to learn nonlinear combinations of technical and sentiment features. LTSM uses data from past days, for instance, the last 10 days of features to predict the next day.  This is similar to a time-series approach, but it requires careful tuning of the overall model. For an MLP, the input is the same as logistic regression, consisting of engineered features like volatility, volume, and horizon returns. For LTSM, the input is based on data from past days, for example, using a 10-day window. In this, each training data consists of features of the past 10 days to predict the next day. 

Backpropagation-style neural networks have been used previously to predict daily stock movements.  Recently, LTSM models have been used to make predictions in large-scale settings. For instance, Fischer and Krauss utilised LTSMs to detect directional movements in the S&P 500 over a long period, from 1992 to 2015, and compared their performance with other models, such as logistic regression and deep networks. In this project's case, this technique can be applied to train on daily engineered indicators. LTSM would use rolling sequences, such as data from the last 10 training days, to predict the direction of the next training day. The expected output would be the same, predicting an up or a down label for the next trading day.

The most common problem with neural networks is overfitting of data, as deep learning typically requires more extensive datasets. It takes longer to set up, and tuning might be needed. The main strength of the model is that it can understand nonlinear effects. However, overfitting remains a concern with limited financial data, and results might be unstable without careful validation. 

Overall, Neural networks can be helpful in prediction when there is substantial data and careful validation is required. In this project, with just a few years of data from a single stock and tweets, simpler models, such as logistic regression, can be more stable. 

Random forest is an ensemble learning technique that builds many decision trees and combines their predictions. Introduced by Breiman in 2001, it is widely used in stock prediction due to its ability to model nonlinear relationships and feature interactions. In 2016, Khaidem et al applied random forests and found that ensemble methods can be effective for this type of classification task. They also achieved high accuracy in long-term predictions across various datasets. This evidence suggests that ensemble tree methods can perform well in classification tasks, such as stock market prediction. 

The input required to use Random Forest would be the same structured daily dataset that I have used, one row per trading day. The input features can include both technical and sentiment indicators, such as multi-horizon returns (1/3/5/12), rolling volatility, moving average gap/crossover, and volume change/z-score. Sentiment can be added as lagged daily features, which means the sentiment of all tweets per day. Unlike logistic regression, random forest does not require feature scaling; however, preprocessing and clearing any missing values are still necessary. The output would be the predicted label (up or down) for the next trading day, and it can also output the probability based on the number of trees voting up or down. 

In my project, I can train a random forest using the same technical indicators and then add lagged tweet features. This could provide a good indication of whether a nonlinear model can yield more signal from the same indicators compared to logistic regression. 

Random Forests capture nonlinear patterns and interactions between indicators; for instance, they can detect if momentum behaves differently under high volatility. It is also generally robust to all outliers and noisy features. On the downside, it is less interpretable than logistic regression, and the trees can overfit if they are too deep, especially in financial datasets where signals can be very weak. Compared to neural networks, however, it is easier to implement and tune, and it works well on small datasets.Although Random Forest is an excellent option for building a strong linear baseline, I chose logistic regression because it is easier to implement, especially when the focus is on determining if sentiment indicators add any additional incremental value. 

In conclusion, Logistic regression is the fastest to implement and the most interpretable of all these, but it is limited in detecting nonlinear patterns. Neural networks are also suitable; however, they require larger datasets and often tuning as well. Random forest can detect nonlinear relationships; however, it is prone to overfitting. This makes logistic regression an appropriate choice for the prototype.

STOCK DATA PRE PROCESSING AND FEATURE ENGINEERING 

I obtained historical Tesla stock data from Kaggle, and it was cleaned by converting the Date column to datetime, sorting the data chronologically, and filtering it to the 2018–2022 period. Furthermore, I engineered technical indicators such as multi horizon returns, moving averages, volatility, and volume-based features. Missing rows and NaN values were dropped before modelling.



```python
data = pd.read_csv("Tesla_stock_data.csv")
```

```python
data.head()
```

```python
data['Date'] = pd.to_datetime(data['Date'])
```

```python
data.dtypes
```

```python
data = data[
    (data['Date'] >= '2018-01-01') &
    (data['Date'] <= '2022-12-31')
]
```

```python
len(data)
```

```python
data = data.sort_values('Date').reset_index(drop=True)
```

```python
# daily return
data['Return'] = data['Close'].pct_change()
```

```python
# checking daily return
data[['Date', 'Close', 'Return']].head()
```

```python
data['Next_Day_Return'] = data['Return'].shift(-1)
```

```python
# up day or down day
data['Target'] = (data['Next_Day_Return'] > 0).astype(int)
```

```python
# feature engineering, calculating multi horizon return, Ret1/3/5/12 means calculating percentage returns over the past 1,3,5,12 trading days
data["Ret1"]  = data["Close"].pct_change(1) 
data["Ret3"]  = data["Close"].pct_change(3)
data["Ret5"]  = data["Close"].pct_change(5)
data["Ret12"] = data["Close"].pct_change(12)
```

```python
# volatility
data["Vol10"] = data["Ret1"].rolling(10).std()
```

```python
# Feature engieering, calculating moving averages. Calculating roling mean of 10 and 20 trading days
data["MA10"] = data["Close"].rolling(10).mean()
data["MA20"] = data["Close"].rolling(20).mean()
# MA gap determines how far price is above/below the medium-term trend (MA20)
data["MA_Gap"] = (data["Close"] / data["MA20"]) - 1
data["MA_Cross"] = (data["MA10"] > data["MA20"]).astype(int)
```

```python
# feature engineering, calculating volume based indicators
# volchg means day to day change in trading volume and volZ20 means how unusual is todays volume compared to the last 20 days 
if "Volume" in data.columns:
    data["VolChg"] = data["Volume"].pct_change()
    data["VolZ20"] = (data["Volume"] - data["Volume"].rolling(20).mean()) / data["Volume"].rolling(20).std()
else:
    data["VolChg"] = np.nan
    data["VolZ20"] = np.nan
```

```python
data = data.dropna().reset_index(drop=True)
```

```python
len(data)
```

```python
# confirming all missing values are gone
data.isna().sum() 
```

```python
# input features, x
features = ["Ret1","Ret3","Ret5","Ret12","Vol10","MA_Gap","MA_Cross","VolChg","VolZ20"]
X = data[features]
```

```python
# defining target, y
y = data['Target']
```

```python
# split features
X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]
```

TWEETS DATA PRE PROCESSING

I obtained the tweet dataset from Kaggle containing the Date and Tweet text. I standardised the dates to a daily format and scored each tweet with a sentiment value using VADER. I then aggregated the tweets per day by calculating the mean sentiment and the daily tweet count. Before merging with the stock data, I lagged the tweet features by one day to handle weekends and reduce leakage.

```python
tweets = pd.read_csv("tweets.csv") 
```

VADER sentiment analysis was used to convert each tweet into a continous numeric sentiment score. 

```python
# importing vader sentiment tool 
try:
    from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
except ImportError:
    import sys
    !{sys.executable} -m pip -q install vaderSentiment
    from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

```python
analyzer = SentimentIntensityAnalyzer()
```

```python
print(tweets.shape)
print(tweets.head())
```

```python
# Parse tweet timestamps 
tweets["Date"] = pd.to_datetime(tweets["Date"], utc=True, errors="coerce")
tweets = tweets.dropna(subset=["Date"]).copy()

# Converting to daily date
tweets["Date"] = tweets["Date"].dt.tz_convert(None).dt.normalize()
```

```python
# computing VADER sentiment score for each tweet
analyzer = SentimentIntensityAnalyzer()

tweets["sentiment_score"] = tweets["Tweet"].astype(str).apply(
    lambda t: analyzer.polarity_scores(t)["compound"]
)
```

```python
# Aggregating tweets to daily features, Tweet_Sent means is the mean of daily sentiment and Tweet_count is the number of tweets that day
daily_tweets = (
    tweets.groupby("Date")
    .agg(Tweet_Sent=("sentiment_score", "mean"),
         Tweet_Count=("sentiment_score", "size"))
    .reset_index()
)
```

```python
daily_tweets.head()
```

```python
# Alligning tweets to the daily calender, including weekends and filling the missing dates
data["Date"] = pd.to_datetime(data["Date"], errors="coerce").dt.normalize()

start_date = data["Date"].min()
end_date   = data["Date"].max()

daily = daily_tweets.set_index("Date").sort_index()

all_days = pd.date_range(start_date, end_date, freq="D")
daily = daily.reindex(all_days)

daily["Tweet_Sent"]  = daily["Tweet_Sent"].fillna(0)
daily["Tweet_Count"] = daily["Tweet_Count"].fillna(0)

```

```python
# Lagging tweet features by 1 day to reduce leakage and routing the weekend sentiment to monday
daily["Tweet_Sent_lag1"]  = daily["Tweet_Sent"].shift(1)
daily["Tweet_Count_lag1"] = daily["Tweet_Count"].shift(1)

daily = daily.dropna().reset_index().rename(columns={"index": "Date"})

# merging lagged tweet features onto stock trading dates
data2 = data.merge(daily[["Date","Tweet_Sent_lag1","Tweet_Count_lag1"]], on="Date", how="left")
data2[["Tweet_Sent_lag1","Tweet_Count_lag1"]] = data2[["Tweet_Sent_lag1","Tweet_Count_lag1"]].fillna(0)

data2["Tweet_Count_lag1_log"] = np.log1p(data2["Tweet_Count_lag1"])
```

COMBINED LOGISTIC REGRESSION ( TWEETS + STOCK)

```python
# adding lagged tweet features to stock feature set
tweet_features = ["Tweet_Sent_lag1", "Tweet_Count_lag1_log"]
combined_features = features + tweet_features
```

```python
# x is features and y is target for the next day for prediction
X = data2[combined_features]
y = data2["Target"].to_numpy()
```

```python
#  80-20 split used
split_idx = int(len(data2) * 0.8)
X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]
y_train, y_test = y[:split_idx], y[split_idx:]
```

```python
# Scaled features
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s  = scaler.transform(X_test)
```

```python
# Training logistic regression for combined features
lr_combo = LogisticRegression(max_iter=2000)
lr_combo.fit(X_train_s, y_train)
y_pred_combo = lr_combo.predict(X_test_s)

y_test_combo = y_test
y_prob_combo = lr_combo.predict_proba(X_test_s)[:, 1]

print("Stock + Tweets accuracy:", round(accuracy_score(y_test, y_pred_combo), 3))
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred_combo))
print(classification_report(y_test, y_pred_combo))
```

Part D – Software Testing and Evaluation. 

To ensure the correctness of the model, the datasets are sorted by date before returns and rolling indicators are calculated. Targets are for the next day, and the label predicts the UP/DOWN signal for the following day. For calculating the sentiment score, it is imperative to ensure that there is no leakage of tweets. They are aggregated daily and lagged by 1 day before being merged with stock price indicators. The model is evaluated using a time-ordered split, with 80% of the data used for training and 20% for testing. This is done to reflect a real-time forecasting scenario. This project involves a binary classification problem, where the target is 1 (UP) if the next-day return is greater than 0, and 0 (DOWN) otherwise.

Data checks and baseline expectation: 

The test set contains 248 days, with DOWN (0) at 51.2% and UP (1) at 48.8%. The model achieves about 51% accuracy when predicting DOWN for next-day prediction, so even a slight improvement would be realistic here. 

Results:

Model 1 (Stock-only Logistic Regression) achieved an accuracy of 0.536, with a confusion matrix of [[64, 63], [52, 69]], and ROC–AUC of 0.536. The model performs only slightly better than chance, correctly predicting 133 out of 248 days. Model 2 (Stock + Tweets Logistic Regression) achieved an accuracy of 0.540, with a confusion matrix of [[63, 64], [50, 71]], and ROC–AUC of 0.530. Adding tweets helps increase accuracy by approximately one per cent (0.536 → 0.540), resulting in one additional correct prediction. However, the AUC decreases slightly, meaning sentiment did not result in overall improvement. The expected accuracy of such a model is about 0.50–0.52, and the produced result shows a slight improvement, 0.536 for stock only and 0.540 for stock + tweet.

Practical insight: 

Comparing the two confusion matrices shows that true positives (UP correctly predicted) increase from 69 to 71, false negatives (UP predicted as DOWN) decrease from 52 to 50, and downside performance is almost unchanged (64 vs 63 correct DOWN). So, tweets mainly helped the model capture a couple more UP days, but the effect is marginal.

Result analysis and ROC curve interpretation: 

The results suggest that technical indicators provide the primary predictive signal in this case, even when the model with tweets gives higher accuracy. The weak AUC supports the conclusion that next-day prediction is close to random. Both ROC curves lie close to the diagonal, and AUC values around 0.53 are indicative of a weak signal. This means that next-day prediction could be noisy even with a simpler model like Logistic Regression. 

Strengths: 

The model follows a proper time series, and there is a clear comparison between the stock and the stock + tweets. The model is aware of leakage; therefore, there is daily aggregation with a 1-day lag to account for weekend tweets. The Logistic Regression model is interpretable and gives a clear signal. The model’s pipeline is lightweight and fully deterministic. Using evaluation metrics such as confusion matrices and ROC/AUC helps to evaluate the performance of the model beyond accuracy. Logistic Regression coefficients can be inspected, which means we can detect which features are pushing the prediction upward or downward. 

Limitations:

 Next-day direction can be extremely noisy, which is why the performance of this model is close to chance. There is a possibility that many tweets are not representative of investor sentiment. The sentiment signal is also limited to tweets; if more sentiment datasets, such as company-specific news, are added, it would help to analyse this further. The availability of software, such as news-specific datasets with matching dates, is quite limited; hence, I used tweets for this project. There is weak separability in the model. The AUC value of 0.53 indicates weak class separation, meaning the model struggles to distinguish between up and down days. Market behaviour is highly volatile; it is challenging to understand the pattern between information and stock price. Moreover, many tweets may contain irrelevant information, which can cause a change in the sentiment score.

Improvements (future work): 

There could be improvements in this model if more sentiment-based, relevant datasets are added. Trying different prediction horizons, such as a 3-day direction, could reduce noise. Multiple time-based splits can help determine whether results are consistent across different time periods. Trying different approaches, like Random Forest, on the same dataset to see whether non-linear relationships improve the performance of the model. This model does not account trading costs and risks. Currently the model uses a 80-20 train test split, using a walkforward evaluation could help as it would train the model on the past data and test on actual future observations.

Bias and fairness-

My project uses Twitter posts to analyse and evaluate the daily sentiment score. A key limitation here is that these tweets are not fully representative of all the market participants, hence limiting the ability to predict with a great deal of accuracy. 


Accessibility, Inclusion and Ethical Implications-

My model aims to predict the probability of the next day's up or down signal. Computational resource analysis-

Data storage requirements-

Storing every single tweet can consume a significant amount of storage; therefore, a sentiment score is calculated based on the mean of daily tweets, thereby reducing storage and model complexity. 

Reducing environmental impact in production-

I have kept the model extremely lightweight, only including imperative features. I store only aggregated daily sentiment features rather than the full raw tweet text, which reduces storage requirements. Logistic Regression is energy-efficient and can be trained quickly on a CPU compared to deep learning approaches, which typically require more compute and longer training times. In a production setting, computing can be further reduced by caching engineered features and retraining only when necessary, rather than retraining frequently.
